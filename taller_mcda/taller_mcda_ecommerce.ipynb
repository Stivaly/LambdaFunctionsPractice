## Paso 4 – Aceptación tecnológica con TAM

### Actividad 4 – Distinguir dato y constructo

**¿Por qué no sería suficiente escribir directamente `tam = {"A": 9, "B": 8, "C": 7, "D": 9}`?**

- No se sabe de dónde salen esos números: quién los puso, si se midió algo o si es la opinión del equipo.
- TAM tiene dos constructos distintos (PU y PEOU) y ahí se juntan en un solo número, por lo que se pierde información (una plataforma puede ser útil pero difícil de usar).
- Según Davis, PU y PEOU se miden con ítems tipo Likert respondidos por usuarios, no se inventan. Un 9 sin instrumento es un número computable, pero no creíble.

**¿Qué información adicional necesitaría para considerar esos valores creíbles?**

- El cuestionario usado (ítems PU1–PU3 y PEOU1–PEOU3) y la escala (Likert 1 a 5).
- Cuántas personas respondieron y si eran usuarios representativos (clientes reales, no solo gente del equipo).
- La tarea que realizaron antes de responder (ej. una compra de prueba en cada plataforma).
- Fecha de la medición, quién la hizo, y el promedio y la dispersión de cada constructo.

## Paso 5 – Tecnología persuasiva, autonomía y valores

### Actividad 5 – Auditar una técnica persuasiva

Técnica seleccionada: **escasez** ("Quedan 3 unidades").

| Pregunta | Respuesta |
|---|---|
| ¿Qué conducta intenta facilitar? | Que el cliente decida comprar más rápido y no deje el producto para después. |
| ¿Qué evidencia muestra al usuario? | El stock disponible. Para que sea responsable, el número debe venir del inventario real en ese momento. |
| ¿Puede ser rechazada? | Sí, el usuario puede ignorar el mensaje y no comprar; no lo bloquea ni le pone contadores de tiempo falsos. |
| ¿Conserva la autonomía? | Sí, siempre que el dato sea verdadero: el usuario recibe información real y decide él. |
| ¿Qué la convertiría en manipulación? | Mostrar "quedan 3" cuando hay 742 (falsa escasez), temporizadores que se reinician o esconder que el mensaje es automático. Eso es influencia + engaño. |

Con el criterio del taller (persuasión responsable = influencia + veracidad + autonomía + transparencia), la escasez es responsable solo si el stock es real y el usuario puede ignorarla. Este tipo de prácticas es lo que se evalúa en el gate de autonomía.

## Paso 6 – Privacidad y contexto

### Actividad 6 – Evaluar un flujo de datos ("Productos para ti")

Se usa el marco de integridad contextual de Nissenbaum: quién + qué dato + para qué + con quién + bajo qué norma.

1. **¿Qué datos utiliza?** Historial de compras, productos vistos, búsquedas, carrito y posiblemente ubicación o datos del perfil.
2. **¿Quién los recolecta?** La tienda a través de la plataforma de e-commerce; también puede intervenir el proveedor de la plataforma o un servicio externo de recomendaciones.
3. **¿Cuál es la finalidad?** Mostrar productos que podrían interesarle al cliente, para facilitar la compra y aumentar las ventas.
4. **¿Quién recibe la inferencia?** El mismo cliente (ve las recomendaciones) y la tienda. No debería llegar a terceros (ej. vender el perfil a otras empresas).
5. **¿El uso es esperable en el contexto?** Sí, en una tienda online es normal que recomienden productos según lo que compré o vi en esa misma tienda. No sería esperable que usen datos de otros sitios, datos sensibles (ej. salud) o que compartan el perfil con terceros.
6. **¿Qué control posee el usuario?** Debería poder ver por qué se le recomienda algo, desactivar la personalización y borrar su historial, además de estar informado en la política de privacidad.

**Norma aplicable:** Ley 19.628 (vigente) y Ley 21.719, que entra en vigencia el 1 de diciembre de 2026. Como el taller se realiza antes de esa fecha, la 21.719 se considera normativa futura, pero conviene que la plataforma ya se ajuste a ella.

**Conclusión:** el flujo respeta la integridad contextual si los datos se quedan dentro de la tienda y se usan solo para recomendar. Si se comparten con terceros o se usan para otra finalidad, se rompe la norma del contexto.

## Paso 7 – Restricciones éticas y jurídicas

### Actividad 7 – Definir el gate de admisibilidad

$$ADM(a) = Legal(a) \land Autonomia(a) \land Privacidad(a) \land Seguridad(a)$$

Condiciones obligatorias:

- **Legal:** cumple la Ley 19.496 y el Decreto 6/2021 (información clara del precio total, derecho de retracto, etc.).
- **Autonomía:** no usa patrones manipulativos (falsa escasez, cuentas regresivas falsas, casillas premarcadas, dificultar la cancelación).
- **Privacidad:** trata los datos personales según la normativa vigente, con consentimiento y finalidad clara.
- **Seguridad:** protege datos y pagos (HTTPS, PCI-DSS en pagos, gestión de vulnerabilidades).

| Alternativa | Legal | Autonomía | Privacidad | Seguridad | ADM |
|---|---|---|---|---|---|
| A ShopFast | PASS | **FAIL** | PASS | PASS | **FAIL** |
| B TrustCart | PASS | PASS | PASS | PASS | PASS |
| C LeanCommerce | PASS | PASS | PASS | PASS | PASS |
| D ProCommerce | PASS | PASS | PASS | PASS | PASS |

Con los datos de la guía ninguna queda INDETERMINATE, porque los gates vienen como True/False. En un caso real, si falta evidencia (por ejemplo, no hay auditoría de seguridad) la marcaría como INDETERMINATE y no la dejaría pasar hasta tenerla; en el código eso equivale a tratarla como FAIL.

## Paso 10 – Funciones de valor (escala 0–10)

Anclas usadas (las de la guía): TCO \$25.000–\$60.000, latencia 150–600 ms, disponibilidad 99–100 %, TAM 1–5. Mantenibilidad y persuasión ya vienen en 0–10.

```python
def valor_menor_mejor(x, minimo, maximo):
    valor = 10 * (maximo - x) / (maximo - minimo)
    return max(0, min(10, valor))


def valor_mayor_mejor(x, minimo, maximo):
    valor = 10 * (x - minimo) / (maximo - minimo)
    return max(0, min(10, valor))


def calcular_valores(alternativas):
    valores = {}
    for nombre, d in alternativas.items():
        valores[nombre] = {
            "costo": valor_menor_mejor(d["tco"], 25000, 60000),
            "rendimiento": valor_menor_mejor(d["latencia"], 150, 600),
            "disponibilidad": valor_mayor_mejor(d["disponibilidad"], 99.0, 100.0),
            "PU": valor_mayor_mejor(d["PU"], 1, 5),
            "PEOU": valor_mayor_mejor(d["PEOU"], 1, 5),
            "mantenibilidad": d["mantenibilidad"],
            "persuasion_support": d["persuasion_support"]
        }
    return valores


valores = calcular_valores(alternativas)

print("VALORES NORMALIZADOS (0-10)")
for a in valores:
    print(a, {c: round(v, 2) for c, v in valores[a].items()})
```

## Paso 12 – Scoring compensatorio

$$S(a) = \sum_{j=1}^{m} w_j \, v_j(a)$$

```python
def calcular_scores(valores, pesos):
    score = {}
    for nombre in valores:
        score[nombre] = sum(
            pesos[c] * valores[nombre][c]
            for c in pesos
        )
    return score


score = calcular_scores(valores, pesos)

print("SCORING SIN GATES")
ranking = sorted(score.items(), key=lambda t: t[1], reverse=True)
for i, (nombre, s) in enumerate(ranking, 1):
    print(i, nombre, round(s, 3))
```

### Actividad 8 – Detectar el problema del scoring puro

3. **¿Sería correcto compensar una pérdida de autonomía con mejor costo y mayor conversión?** No. Si se permite compensar, se está aceptando que manipular a los clientes está bien siempre que salga más barato, lo que va contra los valores éticos y probablemente contra la ley del consumidor. ShopFast gana justamente por la persuasión (aporta 1.5 al score) y por el costo, que es lo que no debería poder "comprar" la autonomía. Por eso la autonomía se trata como veto y no como un criterio más con peso.

## Paso 15 – Formular el problema de optimización

Variable binaria $x_a = 1$ si se selecciona la alternativa $a$, 0 en otro caso.

$$\max Z = \sum_{a \in A} S(a)\,x_a \quad \text{s.a.} \quad \sum_{a \in A} x_a = 1, \qquad ADM(a) = FAIL \Rightarrow x_a = 0$$

```python
def resolver_pulp(alternativas, score, admisibles):
    modelo = pulp.LpProblem("Seleccion_Plataforma_Ecommerce", pulp.LpMaximize)

    x = pulp.LpVariable.dicts("seleccionar", alternativas.keys(), cat="Binary")

    # Funcion objetivo
    modelo += pulp.lpSum(score[a] * x[a] for a in alternativas)

    # Elegir exactamente una plataforma
    modelo += pulp.lpSum(x[a] for a in alternativas) == 1

    # Vetos no compensatorios
    for a in alternativas:
        if not admisibles[a]:
            modelo += x[a] == 0

    modelo.solve(pulp.PULP_CBC_CMD(msg=False))
    return modelo, x
```

## Paso 16 – Resolver

```python
modelo, x = resolver_pulp(alternativas, score, admisibles)

print("ESTADO:", pulp.LpStatus[modelo.status])

for a in alternativas:
    print(a, "score=", round(score[a], 3), "admisible=", admisibles[a], "x=", x[a].value())

for a in alternativas:
    if x[a].value() == 1:
        seleccionada = a
        print("\nSELECCIONADA:", a, "Score:", round(score[a], 3))
```

### Actividad 9 – Interpretar correctamente "Optimal"

**¿Por qué que PuLP muestre `Optimal` no demuestra que la alternativa sea la "mejor del mundo real"?**

Porque `Optimal` solo significa que PuLP encontró la mejor solución del **modelo** que le dimos. Ese modelo depende de la **evidencia** usada (TCO, latencia, encuestas TAM), que puede estar incompleta o mal medida; de las **preferencias** que nosotros definimos (pesos y anclas); y de las **restricciones** que decidimos incluir (elegir una y los gates). Si cambia cualquiera de esas cosas, el óptimo puede ser otra alternativa. Por lo tanto, `Optimal` demuestra que TrustCart es la mejor dado lo que pusimos en el modelo, no que sea la mejor en la realidad.

## Paso 17 – Trazabilidad del resultado

Permite reconstruir: dato → función de valor → peso → contribución → score.

```python
campo_dato = {
    "costo": "tco", "rendimiento": "latencia",
    "disponibilidad": "disponibilidad", "PU": "PU", "PEOU": "PEOU",
    "mantenibilidad": "mantenibilidad",
    "persuasion_support": "persuasion_support"
}

def mostrar_detalle(alternativas, valores, score, admisibles):
    print("DETALLE")
    for a in alternativas:
        print("\n", a, "| Admisible:", admisibles[a], "| Score:", round(score[a], 3))
        for criterio in pesos:
            contribucion = pesos[criterio] * valores[a][criterio]
            print(f"   {criterio:<19} dato={alternativas[a][campo_dato[criterio]]:<8}"
                  f" valor={valores[a][criterio]:.3f}"
                  f" peso={pesos[criterio]:.2f}"
                  f" contribucion={contribucion:.3f}")

mostrar_detalle(alternativas, valores, score, admisibles)
```

### Verificación: ¿implementamos correctamente el modelo que diseñamos?

```python
def verificar(alternativas, score, admisibles, x, seleccionada):
    suma_x = sum(x[a].value() for a in alternativas)
    print("Pesos suman 1:", abs(sum(pesos.values()) - 1) < 1e-9)
    print("Se elige exactamente una alternativa:", suma_x == 1)
    print("Inadmisibles con x=0:",
          all(x[a].value() == 0 for a in alternativas if not admisibles[a]))
    # el ganador de PuLP debe ser el mismo que el maximo entre admisibles
    mejor_admisible = max([a for a in alternativas if admisibles[a]], key=lambda a: score[a])
    print("PuLP coincide con el maximo admisible:", mejor_admisible == seleccionada)

verificar(alternativas, score, admisibles, x, seleccionada)
```

## Paso 19 – Análisis de sensibilidad

Se varía el peso de PEOU entre 0.10 y 0.30; el cambio se redistribuye proporcionalmente entre los demás criterios para que la suma siga siendo 1. "cambio?" se compara con el ganador usando los pesos base.

```python
def ajustar_peso(base, criterio_objetivo, nuevo_peso):
    restantes = {k: v for k, v in base.items() if k != criterio_objetivo}
    suma_restantes = sum(restantes.values())

    nuevos = {}
    for k, v in base.items():
        if k == criterio_objetivo:
            nuevos[k] = nuevo_peso
        else:
            nuevos[k] = (v / suma_restantes) * (1 - nuevo_peso)
    return nuevos


def sensibilidad(alternativas, valores, admisibles, criterio, lista_pesos):
    score_base = calcular_scores(valores, pesos)
    candidatos = [a for a in alternativas if admisibles[a]]
    ganador_base = max(candidatos, key=lambda a: score_base[a])

    print("peso   ganador          score   2do lugar        diferencia  cambio?")
    for nuevo in lista_pesos:
        w = ajustar_peso(pesos, criterio, nuevo)
        scores_esc = calcular_scores(valores, w)

        ordenados = sorted(candidatos, key=lambda a: scores_esc[a], reverse=True)
        ganador = ordenados[0]
        segundo = ordenados[1]
        dif = scores_esc[ganador] - scores_esc[segundo]

        if ganador != ganador_base:
            cambio = "SI"
        else:
            cambio = "NO"

        print(f"{nuevo:.2f}   {ganador:<16} {scores_esc[ganador]:.3f}   "
              f"{segundo:<16} {dif:.3f}       {cambio}")


sensibilidad(alternativas, valores, admisibles, "PEOU", [0.10, 0.15, 0.20, 0.25, 0.30])
```

### Actividad 11 – Robustez

| Peso PEOU | Alternativa ganadora | ¿Cambió? |
|---|---|---|
| 0.10 | D ProCommerce (7.592) | Sí |
| 0.15 | B TrustCart (7.602) | No |
| 0.20 | B TrustCart (7.685) | No |
| 0.25 | B TrustCart (7.767) | No |
| 0.30 | B TrustCart (7.849) | No |

**Clasificación: Sensible.** Al subir el peso de PEOU, TrustCart se mantiene y aumenta su ventaja, pero con 0.10 gana ProCommerce y con 0.15 la diferencia es solo 0.002 (prácticamente un empate). El cambio de ganador ocurre entre 0.10 y 0.15.

La razón es que TrustCart le gana a ProCommerce solo en costo y en PEOU (9.0 vs 7.75), mientras que ProCommerce es mejor en todo lo demás. Al bajar el peso de PEOU, esos otros criterios pesan más y ProCommerce pasa adelante. Sería recomendable probar también el peso del costo y conseguir mejor evidencia de PEOU, que es el criterio que define el resultado.

## Preguntas de cierre

1. **¿Por qué MCDA no comienza necesariamente por los pesos?** Porque los pesos solo tienen sentido después de saber qué se valora y qué criterios representan esos valores. Si se parte por los pesos se termina ponderando lo que es fácil de medir y no lo que importa.
2. **¿Qué aporta Keeney a la construcción del modelo?** El Value-Focused Thinking: partir por los valores y objetivos y ver las alternativas como medios para lograrlos. En el taller se refleja en el árbol V0 → V1…V4.
3. **¿Qué diferencia existe entre la perspectiva de Roy y una búsqueda de "óptimo absoluto"?** Roy plantea que el modelo ayuda a decidir (decision aid) y se construye junto con los participantes; no existe una mejor alternativa objetiva independiente de los valores y el contexto.
4. **¿Por qué PU y PEOU requieren evidencia empírica?** Porque son percepciones de los usuarios; solo se pueden conocer preguntándoles con un instrumento válido, no suponiéndolas.
5. **¿Qué aporta Fogg y qué limitación ética debe añadirse a su uso?** Fogg explica cómo la tecnología puede influir en el comportamiento (recomendaciones, recordatorios, feedback). La limitación es que la influencia debe ser veraz, transparente y respetar la autonomía; si no, es manipulación.
6. **¿Cómo ayuda Value Sensitive Design a un e-commerce?** Obliga a preguntar qué valores afecta cada funcionalidad (privacidad, autonomía, confianza) y a diseñarla considerando eso desde el inicio, no después.
7. **¿Por qué privacidad puede tratarse como límite axiológico?** Porque una violación de privacidad no se arregla con menor costo o mejor rendimiento, y además es una obligación legal. Por eso va como veto.
8. **¿Qué diferencia existe entre un criterio compensatorio y un veto?** En el compensatorio una debilidad se puede compensar con otra fortaleza (trade-off aceptable). En el veto, si falla, la alternativa queda fuera sin importar su score.
9. **¿Qué demuestra realmente PuLP cuando entrega Optimal?** Que encontró la mejor solución del modelo formulado (scores, restricción y vetos), no que sea la mejor en la realidad.
10. **¿Qué diferencia existe entre verificar y validar?** Verificar es revisar que el modelo esté bien implementado (cálculos, código). Validar es revisar si el modelo representa bien el problema real (criterios, stakeholders, datos).
11. **¿Por qué un análisis de sensibilidad aumenta la credibilidad de la conclusión?** Porque muestra si la recomendación depende de un peso específico o se mantiene con cambios razonables. En el caso base mostró que la elección entre TrustCart y ProCommerce era frágil.
12. **¿Cómo se relacionan Claim, Argument y Evidence?** El claim es lo que se afirma (ej. "TrustCart es justificable"), la evidencia son los datos que lo apoyan (TAM, gates, PuLP, sensibilidad) y el argumento es el razonamiento que explica por qué esa evidencia sostiene el claim.
