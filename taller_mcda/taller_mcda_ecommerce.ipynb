## Actividad 1 – Límites de conocimiento

La organización no tiene toda la información sobre las plataformas, trabaja con evidencia limitada (costos, latencia, disponibilidad, encuestas TAM y una revisión ética/legal). Por eso es importante identificar qué cosas no se saben con certeza y qué se podría hacer para reducir esa incertidumbre:

| No se sabe con certeza | Evidencia para reducirlo |
|---|---|
| Costos de mantenimiento a 3 años | Cotizaciones y contratos del proveedor |
| Cómo se comporta la plataforma con más carga (ej. Cyber Day) | Pruebas de carga, SLA del proveedor |
| Si la persuasión realmente aumenta la conversión | Pruebas A/B en un piloto |

Hay que diferenciar lo que es un dato medido de lo que es una suposición. Por ejemplo, "la latencia medida fue 250 ms" es un dato que se observó, mientras que "la plataforma seguirá siendo rápida durante 3 años" es una suposición, porque depende de cosas que todavía no pasan (más usuarios, cambios en la carga, etc.). Lo mismo ocurre con el TCO: el valor que tenemos es una estimación, no un costo ya pagado.

## Actividad 3 – Árbol axiológico

Siguiendo a Keeney, se parte por los valores antes de mirar las alternativas. El valor superior es **V0: adopción responsable y sostenible de una solución de e-commerce**, que se divide en cuatro dimensiones: V1 valor para el usuario, V2 valor económico, V3 responsabilidad ética/jurídica y V4 sostenibilidad tecnológica. En la tabla se relaciona cada stakeholder con lo que valora y con un objetivo que lo represente:

| Stakeholder | ¿Qué valora? | Objetivo |
|---|---|---|
| Cliente | Comprar fácil y que no lo engañen | Maximizar PU y PEOU (V1) |
| Organización | Rentabilidad y ventas | Minimizar TCO (V2) |
| Equipo tecnológico | Estabilidad y fácil mantenimiento | Maximizar disponibilidad y mantenibilidad (V4) |
| Área legal/compliance | Cumplir la ley | Cumplir normativa de consumidor y datos (V3) |
| Sociedad/regulador | Comercio justo y seguro | Respetar autonomía y seguridad (V3) |

Los objetivos de V1, V2 y V4 se pueden medir y comparar entre plataformas, por lo que pasan a ser criterios con peso. Los de V3 no deberían negociarse, por eso más adelante se usan como gates (restricciones).

## Actividad 4 – Dato y constructo

Escribir `tam = {"A": 9, "B": 8, "C": 7, "D": 9}` no basta porque son números sin respaldo: no se sabe quién los puso ni cómo se obtuvieron, por lo que se pueden calcular pero no son creíbles. Además, TAM tiene dos constructos distintos, la utilidad percibida (PU) y la facilidad de uso percibida (PEOU), y en ese diccionario quedan mezclados en un solo valor, perdiendo información (una plataforma puede ser útil pero difícil de usar).

Según Davis, PU y PEOU se miden preguntándole a los usuarios. Para considerar los valores creíbles se necesitaría:

- el cuestionario usado (ítems PU1–PU3 y PEOU1–PEOU3, escala Likert 1 a 5);
- cuántos usuarios respondieron y si eran representativos de los clientes reales;
- la tarea que hicieron antes de responder (por ejemplo, una compra de prueba);
- la fecha de la medición y quién la hizo.

## Actividad 5 – Técnica persuasiva: escasez ("Quedan 3 unidades")

Elegí la técnica de escasez, que consiste en avisarle al cliente que quedan pocas unidades de un producto. Para revisar si es responsable o manipuladora se usa el criterio del taller: persuasión responsable = influencia + veracidad + autonomía + transparencia.

| Pregunta | Respuesta |
|---|---|
| ¿Qué conducta facilita? | Que el cliente compre sin postergar |
| ¿Qué evidencia muestra? | El stock disponible |
| ¿Puede ser rechazada? | Sí, el usuario puede ignorarlo |
| ¿Conserva la autonomía? | Sí, si el stock es real |
| ¿Qué la haría manipulación? | Mostrar "quedan 3" cuando hay 742 (falsa escasez) |

En conclusión, la escasez es responsable solo si el número que se muestra es real y el cliente puede ignorarlo. Si el dato es falso, pasa a ser manipulación porque se influye en la compra usando un engaño. Este tipo de prácticas es lo que se revisa en el gate de autonomía.

## Actividad 6 – Flujo de datos "Productos para ti"

Según Nissenbaum, no basta con preguntar si se usan datos personales, sino que hay que ver quién los usa, qué datos, para qué, con quién se comparten y bajo qué norma.

1. Datos: historial de compras, productos vistos y búsquedas.
2. Los recolecta: la tienda a través de la plataforma.
3. Finalidad: recomendar productos de interés.
4. Recibe la inferencia: el mismo cliente.
5. ¿Es esperable? Sí, mientras los datos no se compartan con terceros ni se usen para otra cosa.
6. Control del usuario: poder desactivar la personalización y borrar su historial.

El flujo es adecuado mientras los datos se queden dentro de la tienda y se usen solo para recomendar, que es lo que un cliente espera. Si se vendieran a otras empresas o se usaran para otro fin, se rompería la integridad contextual. En cuanto a la norma, en Chile aplica la ley de protección de datos personales vigente; la Ley 21.719 entra en vigencia el 1 de diciembre de 2026, por lo que todavía es normativa futura.

## Actividad 7 – Gate de admisibilidad

Algunos valores no deberían poder compensarse con un buen puntaje en otros criterios. Por eso se definen cuatro condiciones obligatorias, y una alternativa solo es admisible si cumple todas:

ADM = Legal ∧ Autonomía ∧ Privacidad ∧ Seguridad

- **Legal:** cumple la ley del consumidor (Ley 19.496) y el Reglamento de Comercio Electrónico.
- **Autonomía:** no usa técnicas manipulativas como falsa escasez o temporizadores falsos.
- **Privacidad:** trata los datos personales según la normativa y con fines claros.
- **Seguridad:** protege los datos y los pagos de los clientes.

Con los datos del caso, la clasificación queda así:

| Alternativa | Legal | Autonomía | Privacidad | Seguridad |
|---|---|---|---|---|
| ShopFast | PASS | **FAIL** | PASS | PASS |
| TrustCart | PASS | PASS | PASS | PASS |
| LeanCommerce | PASS | PASS | PASS | PASS |
| ProCommerce | PASS | PASS | PASS | PASS |

ShopFast es la única que falla, en autonomía. En este caso ninguna queda INDETERMINATE porque los datos vienen como verdadero/falso, pero si faltara evidencia para algún gate (por ejemplo, no hay auditoría de seguridad) se marcaría INDETERMINATE y se trataría como FAIL hasta poder verificarlo.

## Paso 10 – Funciones de valor

Los criterios están en unidades distintas (pesos, milisegundos, porcentajes, escala Likert), así que no se pueden sumar directamente. Por eso se transforman todos a una escala común de 0 a 10. En el costo y la latencia, menor es mejor; en el resto, mayor es mejor. Se usan las anclas de la guía: TCO entre \$25.000 y \$60.000, latencia entre 150 y 600 ms, disponibilidad entre 99 % y 100 % y TAM entre 1 y 5. La mantenibilidad y la persuasión ya vienen de 0 a 10.

Aquí se ve la diferencia entre hecho y valoración: TCO = \$35.000 es una medición, mientras que transformarlo en un 7.14 es una valoración que depende de las anclas elegidas.

```python
def valor_menor_mejor(x, minimo, maximo):
    valor = 10 * (maximo - x) / (maximo - minimo)
    return max(0, min(10, valor))

def valor_mayor_mejor(x, minimo, maximo):
    valor = 10 * (x - minimo) / (maximo - minimo)
    return max(0, min(10, valor))

def calcular_valores(alternativas):
    valores = {}
    for nombre, d in alternativas.items():
        valores[nombre] = {
            "costo": valor_menor_mejor(d["tco"], 25000, 60000),
            "rendimiento": valor_menor_mejor(d["latencia"], 150, 600),
            "disponibilidad": valor_mayor_mejor(d["disponibilidad"], 99.0, 100.0),
            "PU": valor_mayor_mejor(d["PU"], 1, 5),
            "PEOU": valor_mayor_mejor(d["PEOU"], 1, 5),
            "mantenibilidad": d["mantenibilidad"],
            "persuasion_support": d["persuasion_support"]
        }
    return valores

valores = calcular_valores(alternativas)
```

## Paso 11 – Pesos

Los pesos indican cuánto importa cada criterio compensatorio dentro del score final:

- Costo 0.15: importa, pero no debería ganar solo lo más barato.
- Rendimiento 0.10 y disponibilidad 0.10: todas las opciones están en rangos aceptables.
- PU 0.20 y PEOU 0.20: son lo más importante, si el cliente no acepta la plataforma no compra.
- Mantenibilidad 0.10: importa al equipo TI, pero el cliente no lo ve.
- Persuasión 0.15: se quiere conversión, pero solo responsable.

Los pesos los asigné como analista según lo que valora la organización en el caso. Los pesos son preferencias, no datos: otra persona podría asignarlos distinto, y por eso después se revisa la sensibilidad. En el código se comprueba que sumen 1.

```python
pesos = {
    "costo": 0.15,
    "rendimiento": 0.10,
    "disponibilidad": 0.10,
    "PU": 0.20,
    "PEOU": 0.20,
    "mantenibilidad": 0.10,
    "persuasion_support": 0.15
}

assert abs(sum(pesos.values()) - 1.0) < 1e-9
```

## Paso 12 – Scoring compensatorio

El score de cada alternativa es la suma ponderada de sus valores: S(a) = Σ wⱼ·vⱼ(a). Es compensatorio porque un mal puntaje en un criterio se puede compensar con buenos puntajes en otros. En esta celda todavía no se aplican los gates, para ver qué pasaría si solo se mirara el puntaje.

```python
def calcular_scores(valores, pesos):
    score = {}
    for nombre in valores:
        score[nombre] = sum(pesos[c] * valores[nombre][c] for c in pesos)
    return score

score = calcular_scores(valores, pesos)

print("SCORING SIN GATES")
for nombre, s in score.items():
    print(nombre, round(s, 3))
```

### Actividad 8
1. El score más alto es de ShopFast (7.949).
2. No respeta todos los valores: falla en autonomía.
3. No sería correcto, porque la autonomía del cliente no debería poder "comprarse" con menor costo o más conversión. Por eso se trata como veto.

## Paso 13 – Gate no compensatorio

Ahora se aplica el gate de la actividad 7. Una alternativa es admisible solo si cumple las cuatro condiciones (legal, autonomía, privacidad y seguridad); si falla alguna queda descartada sin importar su score. Los criterios compensatorios son trade-offs que se aceptan, y los gates son límites que no se negocian.

```python
def es_admisible(d):
    return d["legal"] and d["autonomia"] and d["privacidad"] and d["seguridad"]

admisibles = {nombre: es_admisible(d) for nombre, d in alternativas.items()}

print("ADMISIBILIDAD")
for nombre, ok in admisibles.items():
    print(nombre, "PASS" if ok else "FAIL")
```

## Pasos 14 a 16 – Modelo PuLP y resolución

Se usa PuLP (instalado con `pip install pulp`) para plantear la selección como un problema de optimización. Se define una variable binaria x[a] que vale 1 si se elige la alternativa y 0 si no. El objetivo es maximizar Σ S(a)·x[a], con la restricción de elegir exactamente una plataforma, y se fuerza x[a] = 0 para las que no son admisibles.

```python
def resolver(alternativas, score, admisibles):
    modelo = pulp.LpProblem("Seleccion_Plataforma_Ecommerce", pulp.LpMaximize)
    x = pulp.LpVariable.dicts("seleccionar", alternativas.keys(), cat="Binary")

    modelo += pulp.lpSum(score[a] * x[a] for a in alternativas)   # funcion objetivo
    modelo += pulp.lpSum(x[a] for a in alternativas) == 1          # elegir una
    for a in alternativas:                                         # vetos
        if not admisibles[a]:
            modelo += x[a] == 0

    modelo.solve(pulp.PULP_CBC_CMD(msg=False))
    print("ESTADO:", pulp.LpStatus[modelo.status])
    for a in alternativas:
        if x[a].value() == 1:
            print("SELECCIONADA:", a, "Score:", round(score[a], 3))
    return x

x = resolver(alternativas, score, admisibles)
```

### Actividad 9
`Optimal` solo indica que PuLP encontró la mejor solución del **modelo** planteado. Ese resultado depende de la **evidencia** usada, de las **preferencias** (pesos) y de las **restricciones** que pusimos, así que no demuestra que sea la mejor opción en el mundo real.

## Paso 17 – Trazabilidad

Para que el resultado no sea solo un número, se muestra para cada alternativa el valor normalizado de cada criterio y cuánto aporta al score (peso × valor). Así se puede reconstruir el camino dato → valor → peso → contribución → score y ver de dónde sale cada puntaje.

```python
print("DETALLE")
for a in alternativas:
    print("\n", a, "Admisible:", admisibles[a], "Score:", round(score[a], 3))
    for criterio in pesos:
        contribucion = pesos[criterio] * valores[a][criterio]
        print(criterio, "valor=", round(valores[a][criterio], 3), "contribucion=", round(contribucion, 3))
```

## Paso 18 – Verificación

Siguiendo a Sargent, se separa verificar (revisar que el modelo esté bien implementado) de validar (revisar si representa bien el problema real). Esta celda revisa lo básico de la verificación: que los pesos sumen 1, que se elija una sola alternativa y que las no admisibles queden en 0.

```python
print("Pesos suman 1:", abs(sum(pesos.values()) - 1) < 1e-9)
print("Una sola alternativa elegida:", sum(x[a].value() for a in alternativas) == 1)
print("Inadmisibles en 0:", all(x[a].value() == 0 for a in alternativas if not admisibles[a]))
```

### Actividad 10 – Verificación manual (TrustCart)

## Paso 19 – Sensibilidad del peso de PEOU

Como los pesos son preferencias, conviene revisar si el resultado cambia al modificarlos. Se prueba el peso de PEOU entre 0.10 y 0.30; la diferencia se reparte proporcionalmente entre los demás criterios para que la suma siga siendo 1. Para cada caso se muestra qué alternativa admisible gana.

```python
def ajustar_peso(base, criterio_objetivo, nuevo_peso):
    suma_restantes = sum(v for k, v in base.items() if k != criterio_objetivo)
    nuevos = {}
    for k, v in base.items():
        if k == criterio_objetivo:
            nuevos[k] = nuevo_peso
        else:
            nuevos[k] = (v / suma_restantes) * (1 - nuevo_peso)
    return nuevos

def sensibilidad(alternativas, valores, admisibles):
    for nuevo_peou in [0.10, 0.15, 0.20, 0.25, 0.30]:
        w = ajustar_peso(pesos, "PEOU", nuevo_peou)
        scores_escenario = calcular_scores(valores, w)
        candidatos = [a for a in alternativas if admisibles[a]]
        ganador = max(candidatos, key=lambda a: scores_escenario[a])
        print("PEOU=", nuevo_peou, "ganador=", ganador, "score=", round(scores_escenario[ganador], 3))

sensibilidad(alternativas, valores, admisibles)
```

### Actividad 11 – Robustez

Con los resultados de la celda anterior se completa la tabla:

| Peso PEOU | Ganadora | ¿Cambió? |
|---|---|---|
| 0.10 | ProCommerce | Sí |
| 0.15 | TrustCart | No |
| 0.20 | TrustCart | No |
| 0.25 | TrustCart | No |
| 0.30 | TrustCart | No |

La recomendación es **sensible**. Cuando el peso de PEOU sube, TrustCart sigue ganando, pero si baja a 0.10 gana ProCommerce, y con 0.15 TrustCart gana por muy poco (7.602 contra 7.600 aprox.). Esto pasa porque TrustCart le gana a ProCommerce principalmente en facilidad de uso y costo, mientras que ProCommerce es mejor en casi todo lo demás. Entonces la elección depende bastante de cuánta importancia se le dé a la facilidad de uso.
